```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


outParser = StrOutputParser() 
llm = ChatOllama(model="llama3.2:1b")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator and translate input in {language}"},
    {"role": "user", "content": "{query}"},
])

final_prompts = prompts.format_messages(language="Hindi" , query="I love india")

print(llm.invoke(final_prompts).content)
```

```python
chains = prompts | llm | outParser

res = chains.invoke({"language":"Hindi" , "query":"I love you ?"})

print(res)
```
