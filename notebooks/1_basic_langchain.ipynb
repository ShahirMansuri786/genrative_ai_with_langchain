```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2:1b")

res = llm.invoke("Who is the PM of India?")

print(res.content)
```

```python
prompts = [
    ("system", "You are a python developer."),
    ("user", "How to sort the array.")
]

res = llm.invoke(prompts)
print(res.content)
```
